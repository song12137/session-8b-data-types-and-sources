```python
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

r = requests.get(
    url,
    headers={
        "User-Agent": "PopulationTableStudy/1.0 (educational notebook)"
    },
    timeout=20
)
r.raise_for_status()

tables = pd.read_html(StringIO(r.text), flavor="lxml")
df = tables[0]

print(df.head())
```
